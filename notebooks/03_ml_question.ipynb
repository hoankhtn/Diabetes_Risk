**Non-ML Question**
- Sau khi phân tách nguy cơ tiểu đường theo các nhóm tuổi và ngưỡng vòng eo, tỷ lệ mắc bệnh tiểu đường ở nhóm người vượt ngưỡng vòng eo có sự khác biệt có so với nhóm dưới ngưỡng theo từng độ tuổi hay không? (D2 - D5)
- Motivation: Câu hỏi đáng trả lời vì đối với các bác sĩ/nhà nghiên cứu, nếu có thể tìm được mỗi liên hệ giữa nhóm tuổi, ngưỡng vòng eo và tỷ lệ tiểu đường thì có khả năng dự đoán được tỷ lệ trẻ em hoặc người già mắc bệnh tiểu đường và kịp thời điều trị.
